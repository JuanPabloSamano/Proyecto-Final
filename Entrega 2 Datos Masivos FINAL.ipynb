# SEGUNDA ENTREGA · Recomendador musical con PCA

**Entrada:** tres canciones favoritas existentes en el dataset y ejercicio a realizar.  
**Técnica:** `StandardScaler` + PCA (≥90 % de varianza) + distancia euclidiana al centroide de las tres canciones.  
**Salida:** playlist con **al menos 10 canciones y 40 minutos**.

> Los rangos BPM por tipo de ejercicio son reglas orientativas del prototipo, no conclusiones causales del estudio de corredores. La playlist resultante es una lista de recomendaciones; no se crea automáticamente en la cuenta de Spotify.


```python
#Integración de más canciones
import pandas as pd
import zipfile
from pathlib import Path

# Mantener intacto df del análisis original de BPM y df_music de corredores
if "df" not in globals():
    raise RuntimeError("Ejecuta primero las 45 celdas originales para crear el DataFrame df.")

RUTA_ZIP_EXTRA = Path("data") / "tracks_features.zip"
NOMBRE_CSV_EXTRA = "tracks_features.csv"

if not RUTA_ZIP_EXTRA.is_file():
    raise FileNotFoundError(
        f"No encuentro el archivo: {RUTA_ZIP_EXTRA.resolve()}. "
        "Verifica que esté dentro de la carpeta data."
    )

COLUMNAS_CATALOGO = [
    "track_id", "track_name", "artists", "duration_ms", "popularity",
    "tempo", "energy", "danceability", "valence", "acousticness",
    "instrumentalness", "liveness", "speechiness", "loudness"
]
COLUMNAS_NUEVO = [
    "id", "name", "artists", "duration_ms", "tempo", "energy",
    "danceability", "valence", "acousticness", "instrumentalness",
    "liveness", "speechiness", "loudness"
]

# Verificar columnas del dataset original y crear popularidad solo si hace falta.
base_original = df.copy()
for col in COLUMNAS_CATALOGO:
    if col not in base_original.columns:
        if col == "popularity":
            base_original[col] = float("nan")
        else:
            raise ValueError(f"Falta {col} en el dataset original.")

# Leer la tabla directamente desde el ZIP sin extraerla al disco.
with zipfile.ZipFile(RUTA_ZIP_EXTRA, "r") as archivo:
    if NOMBRE_CSV_EXTRA not in archivo.namelist():
        raise ValueError(f"El ZIP no contiene {NOMBRE_CSV_EXTRA}.")
    with archivo.open(NOMBRE_CSV_EXTRA) as tabla:
        df_extra = pd.read_csv(tabla, usecols=COLUMNAS_NUEVO, low_memory=False)

filas_nuevas = len(df_extra)
df_extra = df_extra.rename(columns={"id": "track_id", "name": "track_name"})
df_extra["popularity"] = float("nan")  # Campo que NO existe en este ZIP

# Transformar la notación ['Artista 1', 'Artista 2'] a texto legible.
df_extra["artists"] = (
    df_extra["artists"].astype("string")
    .str.replace(r"^\[\s*'|\s*'\s*\]$", "", regex=True)
    .str.replace("', '", ", ", regex=False)
)

# La primera entrega usa df. El recomendador usará esta copia ampliada.
df_catalogo_recomendador = pd.concat(
    [base_original[COLUMNAS_CATALOGO], df_extra[COLUMNAS_CATALOGO]],
    ignore_index=True
)
del base_original, df_extra

print(f"Registros en el nuevo dataset: {filas_nuevas:,}")
print(f"Registros en el catálogo combinado, antes de limpieza: {len(df_catalogo_recomendador):,}")
print("Datos de corredores: sin cambios. El PCA usará df_catalogo_recomendador.")
```

## Preparación del catálogo
Integra el ZIP adicional, limpia campos, filtra valores no válidos y elimina canciones duplicadas.

```python
#Preparar canciones para recomendar
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Estas columnas SÍ existen en el dataset de Spotify de la primera entrega.
CARACTERISTICAS = [
    "tempo", "energy", "danceability", "valence", "acousticness",
    "instrumentalness", "liveness", "speechiness", "loudness"
]
COLUMNAS_OBLIGATORIAS = [
    "track_id", "track_name", "artists", "duration_ms"
] + CARACTERISTICAS

if "df_catalogo_recomendador" not in globals():
    raise RuntimeError("Ejecuta antes la celda de integración de archive (2).zip.")

faltantes = sorted(set(COLUMNAS_OBLIGATORIAS) - set(df_catalogo_recomendador.columns))
if faltantes:
    raise ValueError(f"Faltan columnas en el catálogo combinado: {faltantes}")

# Usar la COPIA ampliada y no modificar el df original ni df_music.
spotify_rec = df_catalogo_recomendador[COLUMNAS_OBLIGATORIAS + ["popularity"]].copy()
for columna in CARACTERISTICAS + ["duration_ms", "popularity"]:
    spotify_rec[columna] = pd.to_numeric(spotify_rec[columna], errors="coerce")
spotify_rec = spotify_rec.dropna(subset=COLUMNAS_OBLIGATORIAS)
# No eliminar las nuevas canciones por carecer de popularity
spotify_rec["track_id"] = spotify_rec["track_id"].astype(str).str.strip()
spotify_rec["track_name"] = spotify_rec["track_name"].astype(str)
spotify_rec["artists"] = spotify_rec["artists"].astype(str)

# BPM válidos, canciones entre 1 y 15 minutos y enlaces identificables.
spotify_rec = spotify_rec[
    spotify_rec["tempo"].between(1, 250)
    & spotify_rec["duration_ms"].between(60000, 900000)
    & (spotify_rec["track_id"].str.len() >= 10)
].copy()

# Un mismo track_id aparece en varios géneros del dataset
spotify_rec = (
    spotify_rec.sort_values("popularity", ascending=False, na_position="last")
               .drop_duplicates(subset="track_id")
               .reset_index(drop=True)
)
# Mantener los índices consecutivos, son posiciones para seleccionar filas de X_pca.

if len(spotify_rec) < 13:
    raise ValueError("Se necesitan al menos 13 canciones válidas para el recomendador.")

print(f"Canciones únicas válidas para recomendar (ambas fuentes): {len(spotify_rec):,}")
print("Variables de similitud:", CARACTERISTICAS)
spotify_rec[["track_name", "artists", "tempo", "duration_ms"]].head()
```

## PCA y reducción de dimensionalidad
Normaliza las nueve características para darles escalas comparables, ajusta el PCA y muestra el porcentaje de varianza explicado.

```python
# Normalización y PCA
# StandardScaler evita que el BPM o la loudness dominen por su escala.
escalador = StandardScaler()
X_normalizado = escalador.fit_transform(spotify_rec[CARACTERISTICAS])

# Conservamos suficientes componentes para explicar al menos el 90% de la varianza.
modelo_pca = PCA(n_components=0.90, svd_solver="full")
X_pca = modelo_pca.fit_transform(X_normalizado)

for i in range(X_pca.shape[1]):
    spotify_rec[f"PC{i+1}"] = X_pca[:, i]

varianza = pd.DataFrame({
    "componente": [f"PC{i+1}" for i in range(X_pca.shape[1])],
    "varianza_explicada": modelo_pca.explained_variance_ratio_,
    "varianza_acumulada": np.cumsum(modelo_pca.explained_variance_ratio_)
})

print(f"Dimensiones originales: {len(CARACTERISTICAS)}")
print(f"Dimensiones después de PCA: {X_pca.shape[1]}")
print(f"Varianza explicada: {modelo_pca.explained_variance_ratio_.sum():.2%}")
display(varianza)

plt.figure(figsize=(8, 4))
plt.plot(range(1, len(varianza)+1), varianza["varianza_acumulada"], marker="o")
plt.axhline(0.90, linestyle="--", label="Meta: 90%")
plt.xticks(range(1, len(varianza)+1))
plt.xlabel("Número de componentes principales")
plt.ylabel("Varianza explicada acumulada")
plt.title("Reducción de dimensionalidad con PCA")
plt.grid(alpha=0.25)
plt.legend()
plt.show()
```

## Entradas del usuario
Permite buscar las tres canciones dentro del catálogo y elegir el tipo de ejercicio. Los intervalos de BPM son heurísticos.

```python
# Preferencias del usuario
# Rangos ORIENTATIVOS de BPM por actividad
EJERCICIOS = {
    "1": ("Caminata", 90, 119),
    "2": ("Trote suave", 120, 139),
    "3": ("Running moderado", 140, 159),
    "4": ("Running intenso", 160, 179),
    "5": ("Intervalos / alta intensidad", 180, 200),
}

def buscar_canciones(texto, limite=8):
    """Encuentra hasta 8 títulos parecidos dentro del CSV de Spotify."""
    texto = texto.strip()
    if not texto:
        return spotify_rec.iloc[0:0].copy()
    coincidencias = spotify_rec[
        spotify_rec["track_name"].str.contains(texto, case=False, regex=False, na=False)
    ].copy()
    return coincidencias.head(limite)


def elegir_tres_canciones():
    """Pregunta por tres títulos y deja seleccionar la versión correcta."""
    ids = []
    print("Escribe una parte del título de cada una de tus 3 canciones favoritas.")
    while len(ids) < 3:
        texto = input(f"Canción favorita {len(ids)+1}/3: ").strip()
        opciones = buscar_canciones(texto)
        opciones = opciones[~opciones["track_id"].isin(ids)].reset_index(drop=True)
        if opciones.empty:
            print("No se encontró. Prueba con otro título que exista en el dataset.")
            continue
        tabla = opciones[["track_name", "artists", "tempo"]].copy()
        tabla.insert(0, "opcion", range(1, len(tabla)+1))
        display(tabla)
        elegido = input("Número de la opción correcta (0 para buscar otra vez): ").strip()
        if not elegido.isdigit() or not (1 <= int(elegido) <= len(opciones)):
            print("Intenta de nuevo.")
            continue
        cancion = opciones.iloc[int(elegido)-1]
        ids.append(cancion["track_id"])
        print("Agregada:", cancion["track_name"], "-", cancion["artists"])
    return ids


def elegir_ejercicio():
    """Permite seleccionar el ejercicio de hoy."""
    print("\n¿Qué ejercicio vas a realizar hoy?")
    for clave, (nombre, minimo, maximo) in EJERCICIOS.items():
        print(f"{clave}. {nombre} ({minimo}-{maximo} BPM)")
    while True:
        opcion = input("Elige un número del 1 al 5: ").strip()
        if opcion in EJERCICIOS:
            return opcion
        print("Opción inválida. Elige del 1 al 5.")
```

## Recomendación
Calcula el centroide de las tres canciones en el espacio PCA, aplica el filtro de BPM y ordena las candidatas por distancia euclidiana hasta cumplir ambos mínimos.

```python
# Recomendador por distancia en PCA
def recomendar_playlist(ids_favoritas, opcion_ejercicio, minimo_canciones=10, minimo_minutos=40):
    """
    1) Centroide de las 3 canciones favoritas en el espacio PCA.
    2) Candidatas dentro del rango BPM del ejercicio.
    3) Orden por distancia euclidiana al centroide (menor = más similar).
    4) Agrega canciones hasta cumplir AMBOS mínimos.
    """
    if len(ids_favoritas) != 3 or len(set(ids_favoritas)) != 3:
        raise ValueError("Debes seleccionar exactamente 3 canciones diferentes.")
    if opcion_ejercicio not in EJERCICIOS:
        raise ValueError("Ejercicio inválido: elige un número del 1 al 5.")
    if minimo_canciones < 10 or minimo_minutos < 40:
        raise ValueError("La entrega exige al menos 10 canciones y 40 minutos.")

    favoritas = spotify_rec[spotify_rec["track_id"].isin(ids_favoritas)]
    if len(favoritas) != 3:
        raise ValueError("Una canción elegida no se encuentra en el dataset limpio.")

    centroide = X_pca[favoritas.index.to_numpy()].mean(axis=0)
    nombre, bpm_min, bpm_max = EJERCICIOS[opcion_ejercicio]
    objetivo_ms = minimo_minutos * 60 * 1000
    posibles = spotify_rec[~spotify_rec["track_id"].isin(ids_favoritas)]

    # Si hay muy pocas canciones, amplía gradualmente el rango e informa del cambio.
    for margen in (0, 10, 20, 40, 250):
        desde_bpm = max(1, bpm_min - margen)
        hasta_bpm = min(250, bpm_max + margen)
        candidatos = posibles[posibles["tempo"].between(desde_bpm, hasta_bpm)].copy()
        if len(candidatos) >= minimo_canciones and candidatos["duration_ms"].sum() >= objetivo_ms:
            break
    else:
        raise ValueError("El catálogo no tiene suficientes canciones o minutos disponibles.")

    # Distancia en todas las componentes retenidas por PCA (no solo PC1 y PC2).
    candidatos["distancia_pca"] = np.linalg.norm(
        X_pca[candidatos.index.to_numpy()] - centroide, axis=1
    )
    candidatos = candidatos.sort_values("distancia_pca")

    seleccionados, duracion_ms = [], 0
    for indice, cancion in candidatos.iterrows():
        seleccionados.append(indice)
        duracion_ms += int(cancion["duration_ms"])
        if len(seleccionados) >= minimo_canciones and duracion_ms >= objetivo_ms:
            break

    if len(seleccionados) < minimo_canciones or duracion_ms < objetivo_ms:
        raise RuntimeError("No fue posible cumplir ambas condiciones mínimas.")

    playlist = candidatos.loc[seleccionados].copy()
    playlist.insert(0, "orden", range(1, len(playlist)+1))
    playlist["duracion"] = playlist["duration_ms"].apply(
        lambda ms: f"{int(ms//60000)}:{int((ms%60000)//1000):02d}"
    )
    playlist["spotify_url"] = "https://open.spotify.com/track/" + playlist["track_id"]

    info = {
        "ejercicio": nombre,
        "bpm_objetivo": (bpm_min, bpm_max),
        "bpm_utilizado": (desde_bpm, hasta_bpm),
        "ampliacion_bpm": margen,
        "cantidad": len(playlist),
        "minutos_totales": duracion_ms / 60000,
        "centroide": centroide,
    }
    return playlist, info
```

## Resultados y exportación
Comprueba los requisitos, muestra las recomendaciones y exporta un CSV. El gráfico PC1–PC2 es una proyección ilustrativa: las distancias reales se calculan en todas las componentes PCA retenidas.

```python
# Visualización, verificación y archivo CSV
def visualizar_playlist(playlist, info, ids_favoritas):
    """Muestra una proyección SOLO VISUAL en PC1 y PC2."""
    muestra = spotify_rec.sample(n=min(1200, len(spotify_rec)), random_state=42)
    favoritas = spotify_rec[spotify_rec["track_id"].isin(ids_favoritas)]
    pc2 = "PC2" if "PC2" in spotify_rec.columns else None
    if pc2 is None:
        print("El PCA conservó una sola componente; no hay gráfica bidimensional.")
        return

    plt.figure(figsize=(9, 6))
    plt.scatter(muestra["PC1"], muestra["PC2"], alpha=0.15, s=12, label="Catálogo (muestra)")
    plt.scatter(favoritas["PC1"], favoritas["PC2"], s=110, marker="*", label="3 favoritas")
    plt.scatter(playlist["PC1"], playlist["PC2"], s=45, label="Canciones recomendadas")
    plt.scatter(info["centroide"][0], info["centroide"][1], s=100, marker="X", label="Centroide")
    plt.xlabel("Componente principal 1")
    plt.ylabel("Componente principal 2")
    plt.title(f"Similitud musical en PCA: {info['ejercicio']}")
    plt.legend()
    plt.grid(alpha=0.2)
    plt.show()
    print("Nota: las distancias reales se calcularon con TODAS las componentes PCA retenidas.")


def revisar_y_exportar(playlist, info, archivo="playlist_recomendada.csv"):
    # Estas condiciones son OBLIGATORIAS para aprobar el requerimiento.
    assert len(playlist) >= 10, "Hay menos de 10 canciones."
    assert playlist["duration_ms"].sum() >= 40*60*1000, "La playlist dura menos de 40 min."
    assert playlist["track_id"].is_unique, "Hay canciones duplicadas."

    columnas_salida = ["orden", "track_name", "artists", "tempo", "duracion",
                       "distancia_pca", "spotify_url"]
    salida = playlist[columnas_salida].copy()
    salida = salida.rename(columns={
        "track_name": "cancion", "artists": "artista", "tempo": "bpm"
    })
    salida["distancia_pca"] = salida["distancia_pca"].round(3)
    salida.to_csv(archivo, index=False, encoding="utf-8-sig")
    print(f"\nEjercicio: {info['ejercicio']}")
    print(f"BPM objetivo: {info['bpm_objetivo']}")
    print(f"Rango BPM realmente utilizado: {info['bpm_utilizado']}")
    if info["ampliacion_bpm"]:
        print("Nota: se amplió el rango para cumplir los mínimos.")
    print(f"Total de canciones: {info['cantidad']}")
    print(f"Duración total: {info['minutos_totales']:.1f} minutos")
    print(f"Archivo creado: {archivo}")
    display(salida)
    return salida
```

## Interacción con usuario

1. Escribe el título de cada canción y elige una opción cuando aparezcan coincidencias.
2. Elige un ejercicio (1–5). 
3. Se creará `playlist_recomendada.csv` y se dibujará un gráfico.


```python
# Ejecuta esta celda para recibir tu playlist personalizada.
ids_favoritas = elegir_tres_canciones()
opcion_ejercicio = elegir_ejercicio()
playlist, info_playlist = recomendar_playlist(ids_favoritas, opcion_ejercicio)
resultado = revisar_y_exportar(playlist, info_playlist)
visualizar_playlist(playlist, info_playlist, ids_favoritas)
```
