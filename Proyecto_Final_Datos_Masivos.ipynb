```python
df_music["bpm_rango"] = pd.cut(
    df_music["bpm_avg"],
    bins=[0, 120, 140, 160, 180, 250],
    labels=["<120", "120-139", "140-159", "160-179", "180+"]
)
```

```python
running_rangos = df_music.groupby("bpm_rango", observed=True)["spm_avg"].mean()

running_rangos
```

```python
spotify_rangos = df.groupby("bpm_rango", observed=True).size()
```

```python
spotify_rangos = spotify_rangos.reset_index(name="canciones")

running_rangos = running_rangos.reset_index()

df_conectado = pd.merge(
    spotify_rangos,
    running_rangos,
    on="bpm_rango"
)

df_conectado
```

```python
plt.figure(figsize=(8,4))

plt.bar(df_conectado["bpm_rango"], df_conectado["canciones"])

plt.xlabel("Rango de BPM")
plt.ylabel("Cantidad de canciones")
plt.title("Cantidad de canciones de Spotify según BPM")

plt.show()
```

```python
plt.figure(figsize=(8,4))

plt.bar(df_conectado["bpm_rango"], df_conectado["spm_avg"])

plt.xlabel("Rango de BPM")
plt.ylabel("Pasos por minuto promedio")
plt.title("Cadencia de corredores según BPM de la música")

plt.show()
```

```python
df_recomendado = df_conectado[df_conectado["spm_avg"] >= 160]

plt.figure(figsize=(8,4))

plt.bar(
    df_recomendado["bpm_rango"],
    df_recomendado["canciones"]
)

plt.xlabel("Rango de BPM")
plt.ylabel("Canciones disponibles en Spotify")

plt.title("Canciones disponibles en rangos de BPM asociados con mayor cadencia")

plt.show()
```
