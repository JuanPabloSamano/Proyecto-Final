```python
import matplotlib.pyplot as plt

corr = df.select_dtypes(include="number").corr()

plt.figure(figsize=(10,8))

plt.imshow(corr)

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation between musical features and popularity")

plt.show()
```

```python
sample = df.sample(3000, random_state=42)

plt.figure(figsize=(8,5))

plt.scatter(
    sample["tempo"],
    sample["popularity"],
    alpha=0.3
)

plt.xlabel("Tempo (BPM)")
plt.ylabel("Popularity")
plt.title("Relationship between Tempo and Song Popularity")

plt.show()
```

```python
import matplotlib.pyplot as plt

# Crear grupos de danceability
df["danceability_group"] = pd.cut(
    df["danceability"],
    bins=3,
    labels=["Baja", "Media", "Alta"]
)

# Calcular promedio de popularidad
dance_popularity = df.groupby(
    "danceability_group",
    observed=True
)["popularity"].mean()


plt.figure(figsize=(8,4))

dance_popularity.plot(
    kind="bar",
    edgecolor="black"
)

plt.xlabel("Nivel de Danceability")
plt.ylabel("Popularidad promedio")

plt.title("Popularidad promedio según Danceability")

plt.xticks(rotation=0)

plt.show()
```

```python
df["tempo_group"] = pd.cut(
    df["tempo"],
    bins=3,
    labels=["Bajo", "Medio", "Alto"]
)

tempo_popularity = df.groupby(
    "tempo_group",
    observed=True
)["popularity"].mean()


tempo_popularity.plot(
    kind="bar",
    figsize=(8,4),
    edgecolor="black"
)

plt.xlabel("Nivel de Tempo")
plt.ylabel("Popularidad promedio")
plt.title("Popularidad promedio según Tempo")

plt.xticks(rotation=0)

plt.show()
```

```python
spotify_data = []

for track_id in df["track_id"].head(300):

    try:
        track = sp.track(track_id)

        spotify_data.append({
            "track_id": track_id,
            "spotify_name": track["name"],
            "artist_api": track["artists"][0]["name"],
            "album_api": track["album"]["name"],
            "spotify_url": track["external_urls"]["spotify"]
        })

    except:
        pass


df_spotify = pd.DataFrame(spotify_data)

df_spotify.head()
```

```python
df_final = df.merge(
    df_spotify,
    on="track_id",
    how="inner"
)

df_final.head()
```
