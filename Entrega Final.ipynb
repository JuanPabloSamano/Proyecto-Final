```python

# =====================================================
# MODELO DE CLASIFICACIÓN SUPERVISADA - SVM
# =====================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import GroupShuffleSplit
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    ConfusionMatrixDisplay
)

# 1. Preparar los datos
VARIABLES_SVM = [
    "bpm_avg",
    "spm_avg",
    "silence_spm_avg",
    "pctDiffWithSilence",
    "phase_avg",
    "res_vec_length",
    "relPhaseAngleDeg",
    "PctChange"
]

VARIABLE_OBJETIVO = "condition"
VARIABLE_GRUPO = "User_ID"

# Comprobar que existen las columnas necesarias
columnas_necesarias = (
    VARIABLES_SVM + [VARIABLE_OBJETIVO, VARIABLE_GRUPO]
)

faltantes = [
    c for c in columnas_necesarias
    if c not in df_music.columns
]

if faltantes:
    raise ValueError(f"Faltan columnas para SVM: {faltantes}")

# Crear copia sin modificar el dataset original
df_svm = df_music[columnas_necesarias].copy()

# Convertir predictores a formato numérico
for columna in VARIABLES_SVM:
    df_svm[columna] = pd.to_numeric(
        df_svm[columna], errors="coerce"
    )

# Eliminar registros incompletos
df_svm = df_svm.dropna(
    subset=columnas_necesarias
).copy()

X = df_svm[VARIABLES_SVM]
y = df_svm[VARIABLE_OBJETIVO].astype(str)
grupos = df_svm[VARIABLE_GRUPO]

print("PREPARACIÓN DE DATOS TERMINADA")
print(f"Registros disponibles: {len(df_svm):,}")
print(f"Características predictoras: {len(VARIABLES_SVM)}")
print(f"Condiciones experimentales: {y.nunique()}")
print("\nDistribución de condiciones:")
display(y.value_counts().sort_index().to_frame("registros"))


# 2. Dividir entrenamiento y prueba
# Se separan participantes completos para evitar que
# sus mediciones aparezcan en ambos conjuntos.

separador = GroupShuffleSplit(
    n_splits=1,
    test_size=0.25,
    random_state=42
)

indices_train, indices_test = next(
    separador.split(X, y, groups=grupos)
)

X_train = X.iloc[indices_train]
X_test = X.iloc[indices_test]

y_train = y.iloc[indices_train]
y_test = y.iloc[indices_test]

grupos_train = set(grupos.iloc[indices_train])
grupos_test = set(grupos.iloc[indices_test])

assert grupos_train.isdisjoint(grupos_test)

if set(y_train.unique()) != set(y_test.unique()):
    raise ValueError(
        "La división no contiene todas las condiciones "
        "en entrenamiento y prueba. Se debe ajustar la división."
    )

print("\nDIVISIÓN DE DATOS")
print(f"Entrenamiento: {len(X_train):,} registros")
print(f"Prueba: {len(X_test):,} registros")
print(f"Participantes entrenamiento: {len(grupos_train)}")
print(f"Participantes prueba: {len(grupos_test)}")


# 3. Crear y entrenar el modelo SVM

modelo_svm = Pipeline([
    ("escalador", StandardScaler()),
    ("clasificador", SVC(
        kernel="rbf",
        C=1.0,
        gamma="scale",
        class_weight="balanced"
    ))
])

modelo_svm.fit(X_train, y_train)

predicciones_svm = modelo_svm.predict(X_test)

print("\nENTRENAMIENTO SVM TERMINADO")


# 4. Crear modelo base para comparación

modelo_base = DummyClassifier(
    strategy="most_frequent"
)

modelo_base.fit(X_train, y_train)
predicciones_base = modelo_base.predict(X_test)


# 5. Evaluar resultados

resultados_svm = pd.DataFrame({
    "Métrica": [
        "Accuracy",
        "Balanced Accuracy",
        "F1 Macro"
    ],
    "Modelo base": [
        accuracy_score(y_test, predicciones_base),
        balanced_accuracy_score(y_test, predicciones_base),
        f1_score(
            y_test, predicciones_base,
            average="macro", zero_division=0
        )
    ],
    "SVM": [
        accuracy_score(y_test, predicciones_svm),
        balanced_accuracy_score(y_test, predicciones_svm),
        f1_score(
            y_test, predicciones_svm,
            average="macro", zero_division=0
        )
    ]
})

print("\nCOMPARACIÓN DE MODELOS")
display(resultados_svm.round(4))

print("\nREPORTE DE CLASIFICACIÓN SVM")
print(
    classification_report(
        y_test,
        predicciones_svm,
        zero_division=0
    )
)


# 6. Matriz de confusión

fig, ax = plt.subplots(figsize=(12, 9))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    predicciones_svm,
    xticks_rotation=45,
    cmap="Blues",
    ax=ax,
    colorbar=False
)

plt.title("Matriz de confusión - Clasificador SVM")
plt.xlabel("Condición predicha")
plt.ylabel("Condición real")
plt.tight_layout()
plt.show()


# 7. Resumen final

print("====================================")
print("MODELO SVM FINALIZADO")
print("====================================")
print(f"Accuracy: {accuracy_score(y_test, predicciones_svm):.2%}")
print(
    f"Balanced Accuracy: "
    f"{balanced_accuracy_score(y_test, predicciones_svm):.2%}"
)
print(
    f"F1 Macro: "
    f"{f1_score(y_test, predicciones_svm, average='macro'):.2%}"
)
print("Kernel utilizado: RBF")
print("Normalización: StandardScaler")
print("Validación: participantes independientes")
```

```python

# =====================================================
# OPTIMIZACIÓN DEL MODELO SVM
# =====================================================

from sklearn.model_selection import (
    GridSearchCV,
    StratifiedGroupKFold
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    ConfusionMatrixDisplay
)

# 1. Preparar los grupos de entrenamiento
grupos_cv = grupos.iloc[indices_train]

# Validación cruzada respetando participantes
validacion = StratifiedGroupKFold(
    n_splits=4,
    shuffle=True,
    random_state=42
)

# 2. Crear pipeline
pipeline_svm = Pipeline([
    ("escalador", StandardScaler()),
    ("clasificador", SVC())
])

# 3. Configurar parámetros a probar
parametros = [
    {
        "clasificador__kernel": ["rbf"],
        "clasificador__C": [0.1, 1, 10, 100],
        "clasificador__gamma": ["scale", "auto", 0.01],
        "clasificador__class_weight": [None, "balanced"]
    },
    {
        "clasificador__kernel": ["linear"],
        "clasificador__C": [0.1, 1, 10, 100],
        "clasificador__class_weight": [None, "balanced"]
    }
]

# 4. Búsqueda de la mejor configuración
busqueda_svm = GridSearchCV(
    estimator=pipeline_svm,
    param_grid=parametros,
    scoring="f1_macro",
    cv=validacion,
    n_jobs=-1,
    refit=True,
    verbose=1
)

print("INICIANDO OPTIMIZACIÓN SVM...")

busqueda_svm.fit(
    X_train,
    y_train,
    groups=grupos_cv
)

# 5. Recuperar el mejor modelo
modelo_svm_optimizado = busqueda_svm.best_estimator_

print("\nOPTIMIZACIÓN TERMINADA")
print("Mejores parámetros:")
print(busqueda_svm.best_params_)

print(
    f"\nMejor F1 Macro en validación cruzada: "
    f"{busqueda_svm.best_score_:.2%}"
)

# 6. Evaluar con los mismos datos de prueba
predicciones_optimizado = modelo_svm_optimizado.predict(X_test)

# 7. Comparar los tres modelos
comparacion_final = pd.DataFrame({
    "Métrica": [
        "Accuracy",
        "Balanced Accuracy",
        "F1 Macro"
    ],
    "Modelo base": [
        accuracy_score(y_test, predicciones_base),
        balanced_accuracy_score(y_test, predicciones_base),
        f1_score(y_test, predicciones_base, average="macro")
    ],
    "SVM original": [
        accuracy_score(y_test, predicciones_svm),
        balanced_accuracy_score(y_test, predicciones_svm),
        f1_score(y_test, predicciones_svm, average="macro")
    ],
    "SVM optimizado": [
        accuracy_score(y_test, predicciones_optimizado),
        balanced_accuracy_score(y_test, predicciones_optimizado),
        f1_score(y_test, predicciones_optimizado, average="macro")
    ]
})

print("\nCOMPARACIÓN FINAL DE MODELOS")
display(comparacion_final.round(4))

# 8. Reporte de clasificación
print("\nREPORTE SVM OPTIMIZADO")

print(
    classification_report(
        y_test,
        predicciones_optimizado,
        zero_division=0
    )
)

# 9. Matriz de confusión optimizada
fig, ax = plt.subplots(figsize=(12, 9))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    predicciones_optimizado,
    xticks_rotation=45,
    cmap="Greens",
    ax=ax,
    colorbar=False
)

plt.title("Matriz de confusión - SVM optimizado")
plt.xlabel("Condición predicha")
plt.ylabel("Condición real")
plt.tight_layout()
plt.show()

# 10. Resumen
print("======================================")
print("SVM OPTIMIZADO FINALIZADO")
print("======================================")

print(
    f"Accuracy: "
    f"{accuracy_score(y_test, predicciones_optimizado):.2%}"
)

print(
    f"Balanced Accuracy: "
    f"{balanced_accuracy_score(y_test, predicciones_optimizado):.2%}"
)

print(
    f"F1 Macro: "
    f"{f1_score(y_test, predicciones_optimizado, average='macro'):.2%}"
)
```

```python

# Integración de Gold de Databricks con el recomendador
import pandas as pd
from databricks import sql
from getpass import getpass

HOST = "dbc-b65a0eb0-f970.cloud.databricks.com"
HTTP_PATH = "/sql/1.0/warehouses/a3d40c78898d88bf"

COLUMNAS_CATALOGO = [
    "track_id", "track_name", "artists", "duration_ms", "popularity",
    "tempo", "energy", "danceability", "valence", "acousticness",
    "instrumentalness", "liveness", "speechiness", "loudness"
]

TOKEN = getpass("Pega tu token de Databricks: ")

consulta = """
SELECT
    track_id,
    track_name,
    artists,
    duration_ms,
    popularity,
    tempo,
    energy,
    danceability,
    valence,
    acousticness,
    instrumentalness,
    liveness,
    speechiness,
    loudness
FROM workspace.default.gold_catalogo_recomendador
"""

# Leer el catálogo completo por lotes para controlar el uso de memoria.
lotes = []

with sql.connect(
    server_hostname=HOST,
    http_path=HTTP_PATH,
    access_token=TOKEN
) as conexion:
    with conexion.cursor() as cursor:
        cursor.execute(consulta)

        while True:
            filas = cursor.fetchmany(50000)
            if not filas:
                break

            lotes.append(pd.DataFrame(
                filas,
                columns=COLUMNAS_CATALOGO
            ))

df_catalogo_recomendador = pd.concat(
    lotes,
    ignore_index=True
)

del lotes

# Conservar el formato que espera el recomendador.
for columna in [
    "duration_ms", "popularity", "tempo", "energy",
    "danceability", "valence", "acousticness",
    "instrumentalness", "liveness", "speechiness", "loudness"
]:
    df_catalogo_recomendador[columna] = pd.to_numeric(
        df_catalogo_recomendador[columna],
        errors="coerce"
    )

for columna in ["track_id", "track_name", "artists"]:
    df_catalogo_recomendador[columna] = (
        df_catalogo_recomendador[columna].astype("string")
    )

print("CATÁLOGO GOLD CARGADO DESDE DATABRICKS")
print(f"Canciones cargadas: {len(df_catalogo_recomendador):,}")
print(f"Columnas: {len(df_catalogo_recomendador.columns)}")

display(df_catalogo_recomendador.head())

```

## Preparación del catálogo
Integra el ZIP adicional, limpia campos, filtra valores no válidos y elimina canciones duplicadas.

```python
#Preparar canciones para recomendar
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Estas columnas SÍ existen en el dataset de Spotify de la primera entrega.
CARACTERISTICAS = [
    "tempo", "energy", "danceability", "valence", "acousticness",
    "instrumentalness", "liveness", "speechiness", "loudness"
]
COLUMNAS_OBLIGATORIAS = [
    "track_id", "track_name", "artists", "duration_ms"
] + CARACTERISTICAS

if "df_catalogo_recomendador" not in globals():
    raise RuntimeError("Ejecuta antes la celda de integración de archive (2).zip.")

faltantes = sorted(set(COLUMNAS_OBLIGATORIAS) - set(df_catalogo_recomendador.columns))
if faltantes:
    raise ValueError(f"Faltan columnas en el catálogo combinado: {faltantes}")

# Usar la COPIA ampliada y no modificar el df original ni df_music.
spotify_rec = df_catalogo_recomendador[COLUMNAS_OBLIGATORIAS + ["popularity"]].copy()
for columna in CARACTERISTICAS + ["duration_ms", "popularity"]:
    spotify_rec[columna] = pd.to_numeric(spotify_rec[columna], errors="coerce")
spotify_rec = spotify_rec.dropna(subset=COLUMNAS_OBLIGATORIAS)
# No eliminar las nuevas canciones por carecer de popularity
spotify_rec["track_id"] = spotify_rec["track_id"].astype(str).str.strip()
spotify_rec["track_name"] = spotify_rec["track_name"].astype(str)
spotify_rec["artists"] = spotify_rec["artists"].astype(str)

# BPM válidos, canciones entre 1 y 15 minutos y enlaces identificables.
spotify_rec = spotify_rec[
    spotify_rec["tempo"].between(1, 250)
    & spotify_rec["duration_ms"].between(60000, 900000)
    & (spotify_rec["track_id"].str.len() >= 10)
].copy()

# Un mismo track_id aparece en varios géneros del dataset
spotify_rec = (
    spotify_rec.sort_values("popularity", ascending=False, na_position="last")
               .drop_duplicates(subset="track_id")
               .reset_index(drop=True)
)
# Mantener los índices consecutivos, son posiciones para seleccionar filas de X_pca.

if len(spotify_rec) < 13:
    raise ValueError("Se necesitan al menos 13 canciones válidas para el recomendador.")

print(f"Canciones únicas válidas para recomendar (ambas fuentes): {len(spotify_rec):,}")
print("Variables de similitud:", CARACTERISTICAS)
spotify_rec[["track_name", "artists", "tempo", "duration_ms"]].head()
```

## PCA y reducción de dimensionalidad
Normaliza las nueve características para darles escalas comparables, ajusta el PCA y muestra el porcentaje de varianza explicado.

```python
# Normalización y PCA
# StandardScaler evita que el BPM o la loudness dominen por su escala.
escalador = StandardScaler()
X_normalizado = escalador.fit_transform(spotify_rec[CARACTERISTICAS])

# Conservamos suficientes componentes para explicar al menos el 90% de la varianza.
modelo_pca = PCA(n_components=0.90, svd_solver="full")
X_pca = modelo_pca.fit_transform(X_normalizado)

for i in range(X_pca.shape[1]):
    spotify_rec[f"PC{i+1}"] = X_pca[:, i]

varianza = pd.DataFrame({
    "componente": [f"PC{i+1}" for i in range(X_pca.shape[1])],
    "varianza_explicada": modelo_pca.explained_variance_ratio_,
    "varianza_acumulada": np.cumsum(modelo_pca.explained_variance_ratio_)
})

print(f"Dimensiones originales: {len(CARACTERISTICAS)}")
print(f"Dimensiones después de PCA: {X_pca.shape[1]}")
print(f"Varianza explicada: {modelo_pca.explained_variance_ratio_.sum():.2%}")
display(varianza)

plt.figure(figsize=(8, 4))
plt.plot(range(1, len(varianza)+1), varianza["varianza_acumulada"], marker="o")
plt.axhline(0.90, linestyle="--", label="Meta: 90%")
plt.xticks(range(1, len(varianza)+1))
plt.xlabel("Número de componentes principales")
plt.ylabel("Varianza explicada acumulada")
plt.title("Reducción de dimensionalidad con PCA")
plt.grid(alpha=0.25)
plt.legend()
plt.show()
```

```python

# Interpretar qué características contribuyen más a cada componente

pesos_pca = pd.DataFrame(
    modelo_pca.components_.T,
    index=CARACTERISTICAS,
    columns=[
        f"PC{i+1}"
        for i in range(modelo_pca.n_components_)
    ]
)

print("Pesos de las características en cada componente:")
display(pesos_pca.round(3))

# Mostrar las 3 características con mayor peso absoluto en cada componente
for componente in pesos_pca.columns:
    principales = pesos_pca[componente].abs().sort_values(
        ascending=False
    ).head(3)

    print(f"\n{componente}: características con mayor peso")
    for variable in principales.index:
        peso = pesos_pca.loc[variable, componente]
        print(f"- {variable}: {peso:.3f}")
```

## Entradas del usuario
Permite buscar las tres canciones dentro del catálogo y elegir el tipo de ejercicio. Los intervalos de BPM son heurísticos.

```python
# Preferencias del usuario
# Rangos ORIENTATIVOS de BPM por actividad
EJERCICIOS = {
    "1": ("Caminata", 90, 119),
    "2": ("Trote suave", 120, 139),
    "3": ("Running moderado", 140, 159),
    "4": ("Running intenso", 160, 179),
    "5": ("Intervalos / alta intensidad", 180, 200),
}

def buscar_canciones(texto, limite=8):
    """Encuentra hasta 8 títulos parecidos dentro del CSV de Spotify."""
    texto = texto.strip()
    if not texto:
        return spotify_rec.iloc[0:0].copy()
    coincidencias = spotify_rec[
        spotify_rec["track_name"].str.contains(texto, case=False, regex=False, na=False)
    ].copy()
    return coincidencias.head(limite)


def elegir_tres_canciones():
    """Pregunta por tres títulos y deja seleccionar la versión correcta."""
    ids = []
    print("Escribe una parte del título de cada una de tus 3 canciones favoritas.")
    while len(ids) < 3:
        texto = input(f"Canción favorita {len(ids)+1}/3: ").strip()
        opciones = buscar_canciones(texto)
        opciones = opciones[~opciones["track_id"].isin(ids)].reset_index(drop=True)
        if opciones.empty:
            print("No se encontró. Prueba con otro título que exista en el dataset.")
            continue
        tabla = opciones[["track_name", "artists", "tempo"]].copy()
        tabla.insert(0, "opcion", range(1, len(tabla)+1))
        display(tabla)
        elegido = input("Número de la opción correcta (0 para buscar otra vez): ").strip()
        if not elegido.isdigit() or not (1 <= int(elegido) <= len(opciones)):
            print("Intenta de nuevo.")
            continue
        cancion = opciones.iloc[int(elegido)-1]
        ids.append(cancion["track_id"])
        print("Agregada:", cancion["track_name"], "-", cancion["artists"])
    return ids


def elegir_ejercicio():
    """Permite seleccionar el ejercicio de hoy."""
    print("\n¿Qué ejercicio vas a realizar hoy?")
    for clave, (nombre, minimo, maximo) in EJERCICIOS.items():
        print(f"{clave}. {nombre} ({minimo}-{maximo} BPM)")
    while True:
        opcion = input("Elige un número del 1 al 5: ").strip()
        if opcion in EJERCICIOS:
            return opcion
        print("Opción inválida. Elige del 1 al 5.")
```

## Recomendación
Calcula el centroide de las tres canciones en el espacio PCA, aplica el filtro de BPM y ordena las candidatas por distancia euclidiana hasta cumplir ambos mínimos.

```python

# Recomendador por similitud coseno ponderada con desempate inteligente

from sklearn.metrics.pairwise import cosine_similarity

# Importancia de cada característica musical.
# BPM conserva la mayor prioridad.
PESOS_CARACTERISTICAS = {
    "tempo": 0.40,
    "energy": 0.15,
    "danceability": 0.15,
    "valence": 0.10,
    "acousticness": 0.05,
    "instrumentalness": 0.05,
    "liveness": 0.03,
    "speechiness": 0.03,
    "loudness": 0.04
}


def recomendar_playlist(
    ids_favoritas,
    opcion_ejercicio,
    minimo_canciones=10,
    minimo_minutos=40
):

    if len(ids_favoritas) != 3 or len(set(ids_favoritas)) != 3:
        raise ValueError("Debes seleccionar exactamente 3 canciones diferentes.")

    if opcion_ejercicio not in EJERCICIOS:
        raise ValueError("Ejercicio inválido: elige un número del 1 al 5.")

    if minimo_canciones < 10 or minimo_minutos < 40:
        raise ValueError("La entrega exige al menos 10 canciones y 40 minutos.")

    if not np.isclose(sum(PESOS_CARACTERISTICAS.values()), 1.0):
        raise ValueError("Los pesos deben sumar 1.")

    favoritas = spotify_rec[
        spotify_rec["track_id"].isin(ids_favoritas)
    ]

    if len(favoritas) != 3:
        raise ValueError("Una canción elegida no se encuentra en el catálogo.")

    # 1. Ponderar las características musicales.
    pesos = np.array([
        PESOS_CARACTERISTICAS[c]
        for c in CARACTERISTICAS
    ])

    factor_pesos = np.sqrt(pesos)
    X_ponderado = X_normalizado * factor_pesos

    # 2. Calcular el centroide de las favoritas.
    indices_favoritas = favoritas.index.to_numpy()

    centroide_ponderado = X_ponderado[
        indices_favoritas
    ].mean(axis=0)

    centroide_pca = X_pca[
        indices_favoritas
    ].mean(axis=0)

    # 3. Obtener el rango de BPM del ejercicio.
    nombre, bpm_min, bpm_max = EJERCICIOS[opcion_ejercicio]
    objetivo_ms = minimo_minutos * 60 * 1000

    posibles = spotify_rec[
        ~spotify_rec["track_id"].isin(ids_favoritas)
    ]

    # 4. Ampliar gradualmente el rango si faltan canciones.
    for margen in (0, 10, 20, 40, 250):

        desde_bpm = max(1, bpm_min - margen)
        hasta_bpm = min(250, bpm_max + margen)

        candidatos = posibles[
            posibles["tempo"].between(desde_bpm, hasta_bpm)
        ].copy()

        if (
            len(candidatos) >= minimo_canciones
            and candidatos["duration_ms"].sum() >= objetivo_ms
        ):
            break

    else:
        raise ValueError(
            "El catálogo no tiene suficientes canciones o minutos disponibles."
        )

    # 5. Calcular similitud coseno ponderada.
    indices_candidatos = candidatos.index.to_numpy()

    similitudes = cosine_similarity(
        X_ponderado[indices_candidatos],
        centroide_ponderado.reshape(1, -1)
    ).ravel()

    candidatos["similitud_coseno"] = similitudes

    # 6. Calcular criterios para resolver empates.

    bpm_promedio_favoritas = favoritas["tempo"].mean()

    # Priorizar canciones dentro del rango original del ejercicio.
    candidatos["diferencia_bpm_objetivo"] = np.where(
        candidatos["tempo"] < bpm_min,
        bpm_min - candidatos["tempo"],
        np.where(
            candidatos["tempo"] > bpm_max,
            candidatos["tempo"] - bpm_max,
            0
        )
    )

    # Cercanía al BPM promedio de las favoritas.
    candidatos["diferencia_bpm_favoritas"] = (
        candidatos["tempo"] - bpm_promedio_favoritas
    ).abs()

    # Cercanía de energía y danceability.
    candidatos["diferencia_energy"] = (
        candidatos["energy"] - favoritas["energy"].mean()
    ).abs()

    candidatos["diferencia_danceability"] = (
        candidatos["danceability"] - favoritas["danceability"].mean()
    ).abs()

    # Preparar popularidad como criterio secundario.
    if "popularity" not in candidatos.columns:
        candidatos["popularity"] = np.nan

    candidatos["popularity"] = pd.to_numeric(
        candidatos["popularity"],
        errors="coerce"
    )

    # 7. Ordenamiento jerárquico y desempate determinista.
    candidatos = candidatos.sort_values(
        by=[
            "similitud_coseno",
            "diferencia_bpm_objetivo",
            "diferencia_bpm_favoritas",
            "diferencia_energy",
            "diferencia_danceability",
            "popularity",
            "track_id"
        ],
        ascending=[
            False, True, True, True, True, False, True
        ],
        na_position="last",
        kind="mergesort"
    )

    # 8. Eliminar canciones repetidas por título y artista.
    candidatos["_nombre_normalizado"] = (
        candidatos["track_name"]
        .astype("string")
        .str.casefold()
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

    candidatos["_artista_normalizado"] = (
        candidatos["artists"]
        .astype("string")
        .str.casefold()
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

    total_antes = len(candidatos)

    candidatos = candidatos.drop_duplicates(
        subset=[
            "_nombre_normalizado",
            "_artista_normalizado"
        ],
        keep="first"
    )

    total_eliminadas = total_antes - len(candidatos)

    candidatos = candidatos.drop(
        columns=[
            "_nombre_normalizado",
            "_artista_normalizado"
        ]
    )

    print(f"Versiones repetidas eliminadas: {total_eliminadas}")

    # 9. Verificar los requisitos después de eliminar duplicados.
    if (
        len(candidatos) < minimo_canciones
        or candidatos["duration_ms"].sum() < objetivo_ms
    ):
        raise ValueError(
            "Después de eliminar duplicados no hay suficientes "
            "canciones o minutos disponibles."
        )

    # 10. Seleccionar canciones hasta cumplir ambos mínimos.
    seleccionados = []
    duracion_ms = 0

    for indice, cancion in candidatos.iterrows():

        seleccionados.append(indice)
        duracion_ms += int(cancion["duration_ms"])

        if (
            len(seleccionados) >= minimo_canciones
            and duracion_ms >= objetivo_ms
        ):
            break

    if (
        len(seleccionados) < minimo_canciones
        or duracion_ms < objetivo_ms
    ):
        raise RuntimeError(
            "No fue posible cumplir ambas condiciones mínimas."
        )

    playlist = candidatos.loc[seleccionados].copy()

    # 11. Preparar la playlist final.
    playlist.insert(
        0,
        "orden",
        range(1, len(playlist) + 1)
    )

    playlist["duracion"] = playlist["duration_ms"].apply(
        lambda ms: (
            f"{int(ms // 60000)}:"
            f"{int((ms % 60000) // 1000):02d}"
        )
    )

    playlist["spotify_url"] = (
        "https://open.spotify.com/track/"
        + playlist["track_id"].astype(str)
    )

    # 12. Información para las gráficas y exportación.
    info = {
        "ejercicio": nombre,
        "bpm_objetivo": (bpm_min, bpm_max),
        "bpm_utilizado": (desde_bpm, hasta_bpm),
        "ampliacion_bpm": margen,
        "cantidad": len(playlist),
        "minutos_totales": duracion_ms / 60000,
        "centroide": centroide_pca,
        "bpm_promedio_favoritas": bpm_promedio_favoritas,
        "pesos": PESOS_CARACTERISTICAS,
        "criterios_desempate": [
            "BPM dentro del rango del ejercicio",
            "BPM cercano a las favoritas",
            "Energía",
            "Danceability",
            "Popularidad",
            "ID de canción"
        ]
    }

    print("RECOMENDADOR FINALIZADO")
    print("Método: similitud coseno ponderada")
    print("Prioridad principal: BPM (40%)")
    print("Desempate: criterios jerárquicos")
    print(f"Canciones recomendadas: {len(playlist)}")

    return playlist, info
```

## Resultados y exportación
Comprueba los requisitos, muestra las recomendaciones y exporta un CSV. El gráfico PC1–PC2 es una proyección ilustrativa: las distancias reales se calculan en todas las componentes PCA retenidas.

```python

# Visualización, verificación y archivo CSV
def visualizar_playlist(playlist, info, ids_favoritas):
    """Muestra una proyección visual de las canciones en PC1 y PC2."""

    muestra = spotify_rec.sample(
        n=min(1200, len(spotify_rec)),
        random_state=42
    )

    favoritas = spotify_rec[
        spotify_rec["track_id"].isin(ids_favoritas)
    ]

    if "PC2" not in spotify_rec.columns:
        print("El PCA conservó una sola componente; no hay gráfica bidimensional.")
        return

    plt.figure(figsize=(9, 6))

    plt.scatter(
        muestra["PC1"], muestra["PC2"],
        alpha=0.15, s=12, label="Catálogo (muestra)"
    )

    plt.scatter(
        favoritas["PC1"], favoritas["PC2"],
        s=110, marker="*", label="3 favoritas"
    )

    plt.scatter(
        playlist["PC1"], playlist["PC2"],
        s=45, label="Canciones recomendadas"
    )

    plt.scatter(
        info["centroide"][0], info["centroide"][1],
        s=100, marker="X", label="Centroide"
    )

    plt.xlabel("Componente principal 1")
    plt.ylabel("Componente principal 2")
    plt.title(f"Recomendaciones por similitud coseno: {info['ejercicio']}")
    plt.legend()
    plt.grid(alpha=0.2)
    plt.show()

    print("La similitud coseno se calculó con las 9 características musicales ponderadas.")
    print("La gráfica muestra una proyección visual en PC1 y PC2.")


def revisar_y_exportar(
    playlist, info, archivo="playlist_recomendada.csv"
):
    # Verificar los requisitos obligatorios.
    assert len(playlist) >= 10, "Hay menos de 10 canciones."
    assert playlist["duration_ms"].sum() >= 40*60*1000, "La playlist dura menos de 40 min."
    assert playlist["track_id"].is_unique, "Hay canciones duplicadas."

    columnas_salida = [
        "orden", "track_name", "artists", "tempo",
        "duracion", "similitud_coseno", "spotify_url"
    ]

    salida = playlist[columnas_salida].copy()

    salida = salida.rename(columns={
        "track_name": "cancion",
        "artists": "artista",
        "tempo": "bpm"
    })

    salida["similitud_coseno"] = salida["similitud_coseno"].round(4)

    salida.to_csv(
        archivo,
        index=False,
        encoding="utf-8-sig"
    )

    print(f"\nEjercicio: {info['ejercicio']}")
    print(f"BPM objetivo: {info['bpm_objetivo']}")
    print(f"Rango BPM utilizado: {info['bpm_utilizado']}")

    if info["ampliacion_bpm"]:
        print("Nota: se amplió el rango para cumplir los mínimos.")

    print(f"Total de canciones: {info['cantidad']}")
    print(f"Duración total: {info['minutos_totales']:.1f} minutos")
    print(f"Archivo creado: {archivo}")

    display(salida)

    return salida
```

```python

# CLUSTERING MUSICAL CON K-MEANS

from sklearn.cluster import KMeans

def visualizar_clusters_kmeans(
    playlist,
    ids_favoritas,
    numero_clusters=5,
    muestra_maxima=20000
):
    """
    Agrupa canciones utilizando K-means sobre las componentes
    principales del PCA y visualiza los grupos en PC1 y PC2.
    """

    # Seleccionar una muestra representativa del catálogo.
    n = min(muestra_maxima, len(spotify_rec))

    indices_muestra = np.random.default_rng(42).choice(
        len(spotify_rec),
        size=n,
        replace=False
    )

    # Entrenar K-means con todas las componentes PCA retenidas.
    modelo_kmeans = KMeans(
        n_clusters=numero_clusters,
        random_state=42,
        n_init=10
    )

    modelo_kmeans.fit(X_pca[indices_muestra])

    # Obtener posiciones de favoritas y recomendadas.
    indices_favoritas = spotify_rec.index[
        spotify_rec["track_id"].isin(ids_favoritas)
    ].to_numpy()

    indices_recomendadas = playlist.index.to_numpy()

    # Asignar cada canción a su grupo.
    grupos_muestra = modelo_kmeans.predict(
        X_pca[indices_muestra]
    )

    grupos_favoritas = modelo_kmeans.predict(
        X_pca[indices_favoritas]
    )

    grupos_recomendadas = modelo_kmeans.predict(
        X_pca[indices_recomendadas]
    )

    # Construir gráfica.
    plt.figure(figsize=(12, 8))

    puntos = plt.scatter(
        X_pca[indices_muestra, 0],
        X_pca[indices_muestra, 1],
        c=grupos_muestra,
        cmap="tab10",
        alpha=0.35,
        s=15,
        label="Canciones del catálogo"
    )

    plt.scatter(
        modelo_kmeans.cluster_centers_[:, 0],
        modelo_kmeans.cluster_centers_[:, 1],
        marker="X",
        s=220,
        c="black",
        edgecolors="white",
        linewidths=1.5,
        label="Centroides K-means"
    )

    plt.scatter(
        X_pca[indices_favoritas, 0],
        X_pca[indices_favoritas, 1],
        marker="*",
        s=250,
        c="gold",
        edgecolors="black",
        label="3 canciones favoritas",
        zorder=5
    )

    plt.scatter(
        X_pca[indices_recomendadas, 0],
        X_pca[indices_recomendadas, 1],
        marker="D",
        s=90,
        c="red",
        edgecolors="white",
        label="Canciones recomendadas",
        zorder=4
    )

    plt.xlabel("Componente principal 1")
    plt.ylabel("Componente principal 2")

    plt.title(
        f"Agrupación musical mediante K-means ({numero_clusters} clusters)"
    )

    plt.colorbar(
        puntos,
        label="Grupo musical"
    )

    plt.legend()
    plt.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()

    print("CLUSTERING K-MEANS TERMINADO")
    print(f"Número de clusters: {numero_clusters}")
    print(f"Canciones utilizadas para entrenar: {n:,}")

    print("\nGrupo de las canciones favoritas:")
    for i, grupo in enumerate(grupos_favoritas, start=1):
        print(f"Favorita {i}: Grupo {grupo + 1}")

    print("\nGrupo de las canciones recomendadas:")
    for i, grupo in enumerate(grupos_recomendadas, start=1):
        print(f"Canción recomendada {i}: Grupo {grupo + 1}")

    return modelo_kmeans
```
